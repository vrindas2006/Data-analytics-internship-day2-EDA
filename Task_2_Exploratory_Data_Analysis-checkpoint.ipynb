# Exploratory Data Analysis – Sample Superstore Dataset

## Internship Task 2 – Data Analytics
#This project performs Exploratory Data Analysis (EDA) on the Sample Superstore dataset to identify patterns, relationships, trends, and outliers in the data.



```python

# 1.Importing Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2.Loading Dataset SampleSuperstore.csv
df = pd.read_csv("SampleSuperstore.csv")

# 3.Display the first 5 rows
print("First 5 rows:")
display(df.head())

# 4.Display the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# 5.Display the column names
print("Column Names:")
print(df.columns)

# 6.Display dataset information
print("Dataset Information:")
df.info()

# 7.Display statistical summary
print("Statistical Summary:")
display(df.describe())
```

```python
# 8.Check for missing values
print("Missing Values:")
print(df.isnull().sum())

# 9.Check for duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

```

```python
# 10.HeatMap

numeric_df = df.select_dtypes(include='number')
correlation = numeric_df.corr()
plt.figure(figsize=(8, 5))
sns.heatmap(correlation, annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

### Observation

The heatmap shows the strength and direction of relationships between numerical variables. Sales and Profit have a positive relationship, while Discount shows a negative relationship with Profit.

```python
# 11.Histogram for Sales Distribution

plt.figure(figsize=(10, 5))
sns.histplot(df['Sales'], bins=30)
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

### Observation

Most transactions have relatively low sales values, while a smaller number of transactions have very high sales values. The distribution is right-skewed.

```python
# 12.BoxPlot for  Profit Distribution and Outliers

plt.figure(figsize=(10, 5))
sns.boxplot(x=df['Profit'])
plt.title('Profit Distribution and Outliers')
plt.xlabel('Profit')
plt.show()
```

### Observation

The boxplot shows a wide range of profit values and several potential outliers. Some transactions generate unusually high profits, while others result in losses.

```python
# 13.ScatterPlot for Relationship Between Sales and Profit

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Sales',
    y='Profit'
)
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

### Observation

The scatter plot shows a generally positive relationship between Sales and Profit. However, some high-sales transactions generate low or negative profit, showing that higher sales do not always guarantee higher profitability.

```python
# 14.BarPlot for Sales by Product Category

category_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)
plt.figure(figsize=(8, 5))
sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

### Observation

Technology generates the highest total sales among the three product categories, followed by Furniture and Office Supplies.

# Top 3 Insights

### 1. Sales and Profit Relationship
Sales and Profit show a generally positive relationship, but higher sales do not always guarantee higher profit. Some high-sales transactions generate low or negative profit.

### 2. Category Performance
Technology is the strongest-performing category in terms of total sales and profit, making it an important contributor to overall business performance.

### 3. Impact of Discounts
The correlation analysis indicates that higher discounts can be associated with lower profitability. Therefore, discount levels should be managed carefully to maintain healthy profit margins.
